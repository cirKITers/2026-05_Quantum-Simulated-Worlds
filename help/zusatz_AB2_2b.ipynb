  <h2><i class="fas fa-puzzle-piece" style="font-size:36px"></i> &nbsp; Toffoli-Gate </h2>
  Untersuche das Toffoli-Gate (CCX):
  <ol>
    <li>Erstelle einen Schaltkreis mit 3 Qubits</li>
    <li>Wende Toffoli an (Qubit 0 und 1 als Kontrolle, Qubit 2 als Ziel)</li>
    <li>Teste mit allen 8 möglichen Eingaben</li>
    <li>Erkläre, warum Toffoli ein AND-Gate implementiert</li>
  </ol>
</div>

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

print("Toffoli-Gate (CCX) Wahrheitstabelle:")
print("  Q0 |  Q1 |  Q2 | Ergebnis")
print("-----+-----+-----+---------")

for q0 in [0, 1]:
    for q1 in [0, 1]:
        for q2 in [0, 1]:
            qc = QuantumCircuit(3, 3)
            if q0: qc.x(0)
            if q1: qc.x(1)
            if q2: qc.x(2)
            
            # Toffoli Gate anwenden
            # Dein Code hier: qc.ccx(0, 1, 2)
            
            qc.measure([0, 1, 2], [2, 1, 0])
            
            job = simulator.run(qc, shots=1)
            result = job.result()
            out = list(result.get_counts().keys())[0]
            print(f"  {q0}  |  {q1}  |  {q2}  |   {out}")
```
