  Untersuche, wie Verschränkung durch Messungen zerstört wird:
  <ol>
    <li>Erstelle einen Bell State</li>
    <li>Messe Qubit 0</li>
    <li>Prüfe, ob die Qubits noch verschränkt sind</li>
    <li>Erkläre das Phänomen</li>

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, partial_trace
from qiskit.visualization import plot_bloch_multivector

# Bell State erstellen
qc = QuantumCircuit(2)
# Dein Code hier:


state = Statevector.from_instruction(qc)
print("Verschränkter Zustand:")
display(plot_bloch_multivector(state))

# Nach der Messung von Qubit 0 kollabiert der Zustand
# Simuliere dies, indem du den Zustand nach einer Messung betrachtest

# Wenn wir Qubit 0 messen und 0 erhalten:
qc_messung = QuantumCircuit(2)
# Dein Code hier:


state_nach_messung = Statevector.from_instruction(qc_messung)
print("Zustand nach Messung von Qubit 0 = 0:")
display(plot_bloch_multivector(state_nach_messung))

# Erklärung:
# Nach einer Messung ist der Zustand nicht mehr verschränkt!
# Die Qubits haben jetzt definite Zustände.
```
