  ```python
  import numpy as np

  def create_oracle():
    oracle_type = np.random.choice(['constant', 'balanced'])
    
    if oracle_type == 'constant':
        constant_value = np.random.choice([0, 1])
        def oracle(x):
            return constant_value
    else:
        # Balanced: XOR der Bits
        def oracle(x):
            return x % 2  # oder sum(bin(x)) % 2
    
    return oracle, oracle_type
  ```
