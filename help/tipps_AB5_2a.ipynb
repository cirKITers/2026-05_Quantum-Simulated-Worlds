  Der Diffusion-Operator verstärkt die Amplitude der gesuchten Lösung. Die Diffusion kann, wie das Orakel, mit einem Multi-Controlled Z und Hadamard-Gates implementiert werden. Du brauchst außerdem X-Gates, da du den Zustand All-Zero (also jedes Qubit im Zustand $|0\rangle$ benutzen möchtest).
    
  ```python
  def diffusion(n):
    qc = QuantumCircuit(n)
    
    # H-X-H = HZH (Inversion um Mittelwert)
    qc.h(range(n))
    qc.x(range(n))
    
    # Multi-Controlled Z
    qc.h(n-1)
    qc.mcx(list(range(n-1)), n-1)
    qc.h(n-1)
    
    qc.x(range(n))
    qc.h(range(n))
    
    return qc
  ```

    <p>Eine Grover-Iteration kombiniert Orakel und Diffusion:</p>
    
  ```python  
  def grover_iteration(ziel, n):
    qc = QuantumCircuit(n)
    
    # Initialisierung
    qc.h(range(n))
    
    # Orakel
    oracle = grover_oracle(ziel)
    qc = qc.compose(oracle)
    
    # Diffusion
    diff = diffusion(n)
    qc = qc.compose(diff)
    
    return qc
  ```
