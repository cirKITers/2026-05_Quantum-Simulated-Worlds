<div class="alert alert-info">
  <h2><i class="fas fa-life-ring" style="font-size:36px"></i> &nbsp;  Tipp 1 </h2>
    <details>
    <summary>Hier klicken!</summary>
    Um den Zustand $|+\rangle$ zu erstellen, wende das H-Gate (Hadamard) auf ein Qubit im Zustand $|0\rangle$ an. Die Funktion <code>h()</code> erwartet als Parameter den Index des Qubits auf dem das Gate angewendet werden soll. In userem Fall ist das der Index $0$ (beachte, dass wir bei $0$ zu zählen beginnen).

  ```python
  qc_plus.h(0)  # Hadamard Gate auf Qubit 0
  ```
</details>
</div>

<div class="alert alert-info">
  <h2><i class="fas fa-life-ring" style="font-size:36px"></i> &nbsp;  Tipp 2 </h2>
    <details>
    <summary>Hier klicken!</summary>
    Um den Zustand $|-\rangle$ zu erstellen, wende das H-Gate (Hadamard) auf ein Qubit im Zustand $|1\rangle$ an. Um den Zustand $|1\rangle$ zu erzeugen, benötigen wir das X-Gate, welches mit der Funktion <code>x()</code> aufgerufen wird. Ähnlich wie beim Hadamard-Gate, erwartet die Funktion <code>x()</code> als Parameter den Index des Qubits auf dem das Gate angewendet werden soll. In userem Fall ist das wieder der Index $0$ (beachte, dass wir bei $0$ zu zählen beginnen).

  ```python
  qc_minus.x(0)  # X Gate auf Qubit 0 um Zustand |−⟩ zu erzeugen
  qc_minus.h(0)  # Hadamard Gate auf Qubit 0
  ```
</details>
</div>

<div class="alert alert-success">
  <h2><i class="fas fa-info" style="font-size:36px"></i> &nbsp;  Lösung </h2>
    <details>
    <summary>Hier klicken!</summary>
   
   Die Lösung für $|+\rangle$:

  ```python
  qc_plus.h(0)  # Hadamard Gate
  state_plus = Statevector.from_instruction(qc_plus)
  ```

  Die Lösung für $|-\rangle$:

  ```python
  qc_minus.x(0)  # X-Gate (NOT)
  qc_minus.h(0)  # Hadamard Gate
  state_minus = Statevector.from_instruction(qc_minus)
  ```
</details>
