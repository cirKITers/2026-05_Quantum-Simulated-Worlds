    <p>Beispiel-Code:</p>

  ```python    
  # XOR (Summe): CNOT von Qubit 0 auf Qubit 2
qc.cx(0, 2)

# AND (Carry): Toffoli mit Qubit 0, 1 als Kontrolle und Qubit 3 als Ziel
qc.ccx(0, 1, 3)
  ```
