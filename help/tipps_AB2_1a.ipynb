    <details>
    <summary>Hier klicken!</summary>
    <p>Die Pauli-Gatter rotieren den Qubit-Zustand auf der Bloch-Sphäre:</p>
    
  - **X-Gate**: Rotation um 180° um die x-Achse (Spiegelt |0⟩↔|1⟩)
  - **Y-Gate**: Rotation um 180° um die y-Achse
  - **Z-Gate**: Rotation um 180° um die z-Achse (ändert die Phase)

</details>
