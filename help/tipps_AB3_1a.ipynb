  1. Wende H-Gate auf Qubit 0 an (erzeugt Superposition)
  2. Wende CNOT an mit Qubit 0 als Kontrolle und Qubit 1 als Ziel

  ```python
  qc_bell = QuantumCircuit(2)
  qc_bell.h(0). # Hadamard auf Qubit 0
  qc_bell.cx(0, 1)  # CNOT mit Qubit 0 als Kontrolle
  ```
