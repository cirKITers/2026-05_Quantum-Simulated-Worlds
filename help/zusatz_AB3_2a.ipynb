  4. Erkläre den Unterschied

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_state_city, plot_histogram
from qiskit_aer import AerSimulator

# GHZ-Zustand
qc_ghz = QuantumCircuit(3)
# Dein Code hier:


state_ghz = Statevector.from_instruction(qc_ghz)
print("GHZ-Zustand:")
display(plot_state_city(state_ghz))

# W-State erstellen (schwieriger!)
# Tipp: Beginne mit |100⟩ und wende Rotationen an
qc_w = QuantumCircuit(3)
# Dein Code hier:


state_w = Statevector.from_instruction(qc_w)
print("W-State:")
display(plot_state_city(state_w))

# Messungen vergleichen
qc_ghz.measure_all()
qc_w.measure_all()

simulator = AerSimulator()
job_ghz = simulator.run(qc_ghz, shots=1000)
job_w = simulator.run(qc_w, shots=1000)

print("GHZ Messungen:")
display(plot_histogram(job_ghz.result().get_counts()))

print("W-State Messungen:")
display(plot_histogram(job_w.result().get_counts()))
```
