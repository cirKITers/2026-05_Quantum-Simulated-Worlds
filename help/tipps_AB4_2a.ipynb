  ```python
  def deutsch_jozsa(oracle_type):
    qc = QuantumCircuit(2, 1)
    
    # 1. H-Gates anwenden
    qc.h(0)

    # 2. Hilfs-Qubit auf |-⟩ setzen
    qc.x(1)
    qc.h(1)
    
    # 2. Orakel anwenden
    if oracle_type == 'balanced':
        qc.cx(0, 1)
    
    # 3. H-Gate auf erstes Qubit
    qc.h(0)
    
    # 4. Messen
    qc.measure(0, 0)
    
    return qc
  ```
