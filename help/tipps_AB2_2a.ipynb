    <p>Das CNOT-Gate (Controlled-NOT) hat zwei Qubits:</p>
    <ul>
    <li>Qubit 0: Kontroll-Qubit</li>
    <li>Qubit 1: Ziel-Qubit</li>
    </ul>
    <p>Wenn das Kontroll-Qubit |1⟩ ist, wird ein X-Gate (NOT) auf das Ziel-Qubit angewendet.</p>
</details>
