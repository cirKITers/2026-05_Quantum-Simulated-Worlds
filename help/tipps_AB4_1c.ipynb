  def create_quantum_oracle(oracle_type):
    qc = QuantumCircuit(2)
    if oracle_type == 'constant_0':
        pass  # Nichts tun
    elif oracle_type == 'constant_1':
        qc.x(1)  # Flippe das Ausgabe-Qubit
    elif oracle_type == 'balanced':
        qc.cx(0, 1)  # CNOT
    return qc
