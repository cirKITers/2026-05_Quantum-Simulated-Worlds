   Die kartesischen Koordinaten werden so berechnet:

  ```python
  def bloch_to_cartesian(theta, phi):
    x = np.sin(theta) * np.cos(phi)
    y = np.sin(theta) * np.sin(phi)
    z = np.cos(theta)
    return x, y, z
  ```
