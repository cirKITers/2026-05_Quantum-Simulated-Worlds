    <p>Ein Volladdierer mit 3 Eingängen (a, b, carry_in):</p>
    
  1. Summe = a XOR b XOR carry_in (zwei CNOTs)
  2. Carry_out = (a AND b) OR (carry_in AND (a XOR b))

  <p>Schritt-für-Schritt:</p>
  
  ```python
  # Summe berechnen
  qc.cx(0, 3)  
  qc.cx(1, 3)  # a XOR b -> in Qubit 3
  qc.cx(2, 3)  # (a XOR b) XOR carry_in

  # Carry_out berechnen
  # Erstes AND: a AND b
  qc.ccx(0, 1, 4)  # Ergebnis in Qubit 4
  # Zweites AND: carry_in AND (a XOR b)
  # Dann OR der beiden Ergebnisse
  ```

</details>

  qc.ccx(0, 1, 4)  # a, b → Carry_out
  qc.ccx(0, 2, 4)  # a, carry_in
  qc.ccx(1, 2, 4)  # b, carry_in
