    <p>Die optimale Anzahl an Iterationen ist:</p>
    
  - Bei $N = 2^n$ für $n$ Qubits möglichen Lösungen: $\sqrt{N}$ Iterationen
  - Bei $8$ Elementen ($3$ Qubits): $\sqrt{8} \approx 2,83$ Iterationen
  - Bei $16$ Elementen ($4$ Qubits): $\sqrt{16} = 4$ Iterationen
    
  <p>Zu viele Iterationen können die Erfolgsrate wieder verschlechtern! Probier das auch gern mal aus.</p>
