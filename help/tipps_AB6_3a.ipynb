  def init_input(feld, circuit, register):
    """Funktion, die das Feld als Input auf einem Register initialisiert"""
    for idx, zeile in enumerate(feld):
        for idy, value in enumerate(zeile):
            if value == 1:
                circuit.x(register[idx * len(start_feld) + idy])

<div class="alert alert-info">
  <h2><i class="fas fa-life-ring" style="font-size:36px"></i> &nbsp; Tipp 2 </h2>
    <details>
    <summary>Hier klicken!</summary>
    
  Das Abbilden der Klicks auf dem Quantencomputer funktioniert ähnlich, wie im Klassischen. Auch hier schauen wir, welches Qubit gerade geklickt werden soll und bestimmen dann die Nachbarn. Den eigentlichen Klick erzeugen wir diesmal allerdings nicht selbst, sondern nutzen die Superposition, die alle Klicks für uns bereits enthält. Wir müssen also einen Test hinzufügen, ob das Qubit $1$ ist und wenn ja, dann ändern wir den Wert des entsprechenden Nachbarn. 
  
  Diese Beschreibung klingt für dich nach einem `CX`? Und genau das können wir für die Klicks auch benutzen. Wir müssen also pro Nachbarn ein `CX` auf den Circuit packen. Dafür können wir die bereits implementierte Funktion `nachbarn(x, y)` benutzen. Dafür haben wir euch die Hilfsfunktionen `qubit_zu_koordinate(qubit)` und `koordinate_zu_qubit(x, y)` zur Verfügung gestellt.

</details>
</div>

    """Erstellt ein Orakel für Lights-out"""

    # Abbildung der Klicks
    for i in range(len(suche)):
        x, y = qubit_zu_koordinate(i)
        for nx, ny in nachbarn(x, y):
            circuit.cx(suche[i], data[koordinate_zu_qubit(nx, ny)])
    # Markieren der korrekten Lösung
    circuit.barrier()
    circuit.x(data)
    circuit.h(phase)
    circuit.mcx(data, phase)
    circuit.h(phase)
    circuit.x(data)
    circuit.barrier()
