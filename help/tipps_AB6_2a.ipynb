  Es gibt maximal 4 Nachbarn: oben, unten, links und rechts. Du kannst die Position der Nachbarn relativ beschreiben:

  ```python
  [
    (-1, 0),  # oben
    (1, 0),  # unten
    (0, -1),  # links
    (0, 1),  # rechts
  ]
  ```

  Diese Liste kannst du gleichzeitig benutzen um zu checken, ob es sich tatsächlich um einen **existierenden** Nachbarn handelt.

    <details>
    <summary>Hier klicken!</summary>
  
  Um zu überprüfen, ob ein Nachbar **existiert**, musst du überprüfen, ob du außerhalb des Spielfeldes bist. Also außerhalb der eigentlichen Listendefinition. Das kann bedeuten, dass du versuchst auf einen negativen Index zuzugreifen, oder auf einen, der größer ist als die eigentliche Dimension des Spielfeldes.

  ```python
  0 <= nx < len(feld)
  ```

  Du willst also sicherstellen, dass der neue mögliche Index $\ge 0$ ist, aber auch kleiner als die Spielfeldlänge (da wir immer bei 0 zu zählen beginnen, kannst du auf `len(feld)` testen). Das solltest du sowohl für die x-Richtung, als auch die y-Richtung testen.
</details>
</div>

```python
def nachbarn(x, y, size=len(feld)):
    """Bestimmt die validen Nachbarn von Feld x, y (inklusive x, y)"""
    nachbarn = [[x, y]]

    for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nx, ny = (x + dx), (y + dy)
        if 0 <= nx < size and 0 <= ny < size:
            nachbarn.append([nx, ny])

    return nachbarn
```
