    <details>
    <summary>Hier klicken!</summary>
   Um einen Schaltkreis mit Messung zu erstellen:

  ```python
  qc = QuantumCircuit(1, 1)  # 1 Qubit, 1 klassisches Bit
  qc.h(0)  # Hadamard Gate
  qc.measure(0, 0)  # Messe Qubit 0 in klassisches Bit 0
  ```

  <p>Um den Schaltkreis auszuführen:</p>

  ```python
  simulator = AerSimulator()
  job = simulator.run(qc, shots=1000)
  result = job.result()
  counts = result.get_counts()
  print(counts)
  ```
</details>
