  Bei einem Orakel wird in der Regel auf den Zustand $|1\rangle$ eines Qubits getestet. Das heißt, wenn du nur den Zustand $|101\rangle$ markieren möchtest, musst du das entsprechend vorbereiten. Auf die $1$ kannst du bereits gut testen, bei der $0$ würde der Test allerdings fehl schlagen, d.h. du musst aus der $0$ erst noch eine $1$ machen. Das entsprechende Gate dafür solltest du bereits kennen.

  | Zustand | Test | Gate notwendig? |
  |---------|------|-----------------|
  | 1       | True | nein            |
  | 0       | False| ja              |
  | 1       | True | nein            |

<div class="alert alert-info">
  <h2><i class="fas fa-life-ring" style="font-size:36px"></i> &nbsp; Tipp 4</h2>
    <details>
    <summary>Hier klicken!</summary>

  Ein Multi-Controlled X kannst du folgendermaßen verwenden:

  ```python
  qc.mcx([0, 1], [2])
  ```

  In diesem Beispiel sind die Controls auf den Qubits $0$ und $1$ und das Ziel-Qubit, auf dem das X-Gate angewendet wird, ist das Qubit $2$.

</details>
</div>

<div class="alert alert-info">
  <h2><i class="fas fa-life-ring" style="font-size:36px"></i> &nbsp; Tipp 5</h2>
    <details>
    <summary>Hier klicken!</summary>

  Um aus dem MCX-Gate ein Multi-Control Z Gate zu machen, benötigst du ein H-Gate. Damit kannst du eine Drehung um die Z-Achse ermöglichen:

  ```python
  qc.h(2)
  qc.mcx([0, 1], [2])
  qc.h(2)
  ```

  Nach der Anwendung deines MCX-Gates musst du zusehen, das H-Gate erneut anzuwenden, damit du wieder in der "normalen" Basis unterwegs bist.

</details>
</div>

  ```python
  def grover_oracle(ziel):
    n = len(ziel)
    qc = QuantumCircuit(n)
    
    # X-Gates auf Qubits, die 0 sein sollen
    for i, bit in enumerate(ziel):
        if bit == '0':
            qc.x(i)
    
    # Multi-Controlled Z
    qc.h(n-1)
    qc.mcx(list(range(n-1)), n-1)
    qc.h(n-1)
    
    # X-Gates rückgängig
    for i, bit in enumerate(ziel):
        if bit == '0':
            qc.x(i)
    
    return qc
  ```
