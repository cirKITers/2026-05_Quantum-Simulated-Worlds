  Im Quantencomupting ist es aktuell noch wichtig, so wenige Gatter und so wenige Qubits wie möglich zu verwenden. Aktuell haben wir unseren Volladdierer mit 5 Qubits implementiert. Es ist aber auch möglich, den Volladdierer mit nur 4 Qubits darzustellen.
  <ol>
    <li>Erstelle einen Quantenschaltkreis mit 4 Qubits</li>
    <li>Implementiere die Funktion des Volladdierers</li>
  </ol>

  Um den Volladdierer mit nur 4 Qubits abzubilden, musst du eins der Qubits der Inputqubits wiederverwenden, z.B. Carry_in.
</div>

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Volladdierer erstellen
# Qubit 0: Eingabe a
# Qubit 1: Eingabe b
# Qubit 2: Carry-in und Summe
# Qubit 3: Carry-out

def volladdierer(a, b, carry_in):
    """Führt einen Volladdierer aus"""
    qc = None
    
    # Setze Eingaben
    if a == 1: qc.x(0)
    if b == 1: qc.x(1)
    if carry_in == 1: qc.x(2)
    
    # Dein Code hier:
    # 1. Summe = a XOR b XOR carry_in
    # 2. Carry_out berechnen
    
    qc.measure([2, 3], [1, 0])
    
    simulator = AerSimulator()
    job = simulator.run(qc, shots=1)
    result = job.result()
    counts = result.get_counts()
    
    ergebnis = list(counts.keys())[0]
    summe = int(ergebnis[0])
    carry_out = int(ergebnis[1])
    
    return summe, carry_out

# Teste
print("Volladdierer Test:")
for a in [0, 1]:
    for b in [0, 1]:
        for c_in in [0, 1]:
            s, c_out = volladdierer(a, b, c_in)
            print(f"{a}+{b}+{c_in} = {s}, Carry={c_out}")
```
