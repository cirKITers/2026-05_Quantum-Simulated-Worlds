  Untersuche, wie sich Bell States unter verschiedenen Operationen verhalten:
  
  1. Erstelle den Bell State $|\phi^+\rangle$
  2. Wende X auf Qubit 0 an
  3. Wende H auf Qubit 0 an
  4. Erkläre, wie sich der Zustand verändert

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_state_city

# Bell State |Φ⁺⟩ erstellen
qc = QuantumCircuit(2)
# Dein Code hier:


state = Statevector.from_instruction(qc)
print("Bell State |Φ⁺⟩:")
display(plot_state_city(state))

# Jetzt: X auf Qubit 0 anwenden
qc_x = QuantumCircuit(2)
# Dein Code hier:


state_x = Statevector.from_instruction(qc_x)
print("Nach X auf Qubit 0:")
display(plot_state_city(state_x))

# Jetzt: H auf Qubit 0 anwenden
qc_h = QuantumCircuit(2)
# Dein Code hier:


state_h = Statevector.from_instruction(qc_h)
print("Nach H auf Qubit 0:")
display(plot_state_city(state_h))
```
