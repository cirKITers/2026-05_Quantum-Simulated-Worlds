## Was ist ein Jupyter Notebook?

Ein **Jupyter Notebook** ist eine interaktive Umgebung, in der du Text, Code und Bilder an einem Ort kombinieren kannst. Stell es dir wie ein digitales Arbeitsheft vor!

**Stell dir das so vor:**
- Ein Notizbuch mit vielen Seiten
- Jede Seite kann entweder **Text** (wie in einem Lehrbuch) oder **Code** (wie ein Rezept) enthalten
- Du kannst die Seiten in beliebiger Reihenfolge lesen und bearbeiten
- Am Ende kannst du deine Ergebnisse direkt im Notebook sehen

In diesem Kurs verwenden wir **bwJupyter**, eine spezielle Version von Jupyter, die vom Karlsruher Institut für Technologie (KIT) bereitgestellt wird. Hier ist für dich bereits alles eingerichtet und du wirst wöchentlich neue Notebooks von uns bekommen.

## Aufbau eines Jupyter Notebooks

Ein Jupyter Notebook besteht aus **Zellen**. Es gibt zwei Haupttypen:

### 1. Markdown-Zellen (Text)

Markdown-Zellen enthalten formatierten Text, Bilder und sogar mathematische Formeln. Sie sind wie Seiten in einem Buch – du kannst sie lesen, aber normalerweise nicht direkt bearbeiten.

**Beispiele für Markdown:**
- **Fetter Text** wird mit `**Text**` erstellt
- *Kursiver Text* wird mit `*Text*` erstellt
- Listen werden mit `-` am Anfang erstellt
- Mathematische Formeln wie $E = mc^2$ werden mit `$Formel$` erstellt

### 2. Code-Zellen (Programmiercode)

In Code-Zellen kannst du Python-Code schreiben und ausführen. Die Zelle zeigt dann das Ergebnis direkt darunter an.

**Beispiel:**

```python
# Dies ist eine Code-Zelle!
# Wir können hier Python-Code schreiben und ausführen

name = "Welt"
print(f"Hallo, {name}!")

# Berechnungen sind auch möglich
ergebnis = 2 + 3
print(f"2 + 3 = {ergebnis}")
```

## Wichtige Tastenkombinationen

Hier sind die wichtigsten Tastenkombinationen, die du kennen solltest:

| Tastenkombination | Aktion |
|------------------|--------|
| **Shift + Enter** | Zelle ausführen und zur nächsten Zelle wechseln |
| **Strg + Enter** | Zelle ausführen, aber in der gleichen Zelle bleiben |
| **Esc** | Zur Befehlsmodus wechseln (wenn du nicht mehr in einer Zelle bist) |
| **Enter** | In eine Zelle hineinklicken, um sie zu bearbeiten |

<div class="alert alert-success">
  <h2><i class="fas fa-exclamation-triangle" style="font-size:36px"></i> &nbsp; Wichtig: Markdown neu rendern</h2>
  <p>Wenn du versehentlich eine <strong>Markdown-Zelle bearbeitest</strong> (z.B. weil du doppelt auf den Text geklickt hast), siehst du den formatierten Text nicht mehr – stattdessen siehst du den Rohcode.</p>
  <p><strong>So renderst du die Zelle neu:</strong></p>
  <ol>
    <li>Klicke auf die Zelle (sie sollte blau umrandet sein)</li>
    <li>Drücke <strong>Shift + Enter</strong> oder <strong>Strg + Enter</strong></li>
    <li>Alternativ kannst du auch oben auf das Play-Symbol klicken, auch damit verlässt du den Editiermodus.</li>
    <li>Der Text sollte nun wieder formatiert angezeigt werden!</li>
  </ol>
</div>
