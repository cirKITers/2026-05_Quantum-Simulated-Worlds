## Einleitung: Klassische Bits vs. Qubits

In klassischen Computern arbeiten wir mit **Bits**, die entweder den Wert $0$ oder $1$ haben. Quantencomputer hingegen nutzen **Qubits** (Quantenbits), die sich grundlegend anders verhalten.

**Stell dir das wie ein Licht vor:**
- Ein klassisches Licht ist entweder **AN** ($1$) oder **AUS** ($0$)
- Ein Quantenlicht kann gleichzeitig **AN und AUS** sein – wie ein Licht, das zwischen An und Aus schwebt!

Erst wenn wir "hineinschauen" (messen), entscheidet sich das Licht für einen Zustand.

## Was bedeuten $|0\rangle$ und $|1\rangle$?

In der Quantenphysik verwenden wir eine besondere Schreibweise, die **Braket-Notation** heißt. Sie wurde von Paul Dirac erfunden.

**Der "Ket"** ($|\cdot\rangle$) steht für einen **Zustand** – wie ein Foto von einem bestimmten Moment.

* $|0\rangle$ bedeutet: Das Qubit ist im Zustand "0" (AUS)
* $|1\rangle$ bedeutet: Das Qubit ist im Zustand "1" (AN)

## Superposition: Mehrere Zustände gleichzeitig

Ein Qubit kann sich in einer **Superposition** befinden – es ist also gleichzeitig in beiden Zuständen!

**Beispiel:**
$$|\psi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle$$

Hier siehst du, dass vor jedem Basiszustand ein **Koeffizient** steht. Diesen Koeffizienten nennen wir **Amplitude**. Die Amplitude gibt an, wie "stark" der jeweilige Zustand vertreten ist. In diesem Beispiel ist die Amplitude von $|0\rangle$ gleich $\frac{1}{\sqrt{2}}$ und die Amplitude von $|1\rangle$ ist ebenfalls $\frac{1}{\sqrt{2}}$.

**Wichtig:** Die Wahrscheinlichkeit, bei einer Messung einen bestimmten Zustand zu erhalten, berechnet sich aus dem Quadrat des Betrags der Amplitude:
- $P(|0\rangle)=\left|\frac{1}{\sqrt{2}}\right|^2=\frac{1}{2}=50\%$
- $P(|1\rangle)=\left|\frac{1}{\sqrt{2}}\right|^2=\frac{1}{2}=50\%$

Erst wenn wir messen, "entscheidet" sich das Qubit für einen Zustand.

## Was sind Vektoren? 📐

**Vektoren** sind nichts anderes als **Punkte** in einem Koordinatensystem, zum Beispiel werden sie in einem 2D-Koordinatensystem durch zwei Zahlen beschrieben.

### Ein einfaches Beispiel

Stell dir ein 2D-Koordinatensystem vor:

<img src="../figs/vektoren_koordinatensystem.png" alt="Koordinatensystem mit Punkt P" style="width:500px; margin:auto;">

Ein Punkt in diesem Koordinatensystem wird durch zwei Zahlen beschrieben:
- Die **erste Zahl** sagt, wie weit wir nach **rechts** gehen (x-Wert)
- Die **zweite Zahl** sagt, wie weit wir nach **oben** gehen (y-Wert)

Der Punkt **P** hat die Koordinaten **(0.6, 0.8)**. Das bedeutet:
- Wir gehen 0.6 Einheiten nach rechts
- Wir gehen 0.8 Einheiten nach oben

### Vektoren in Python

In Python verwenden wir **NumPy**, um mit Vektoren zu arbeiten:

```python
import numpy as np

# Punkt P mit Koordinaten (0.6, 0.8)

# Auf die x-Koordinate zugreifen
print(point_p[0])  # Ausgabe: 0.6

# Auf die y-Koordinate zugreifen
print(point_p[1])  # Ausgabe: 0.8
```

## 2) Die Bloch Sphere

Die Bloch Sphere ist eine Kugeloberfläche, auf der wir jeden möglichen Qubit-Zustand als Punkt darstellen können.

Die Pole der Bloch Sphere entsprechen den Basis-Zuständen:
- Nordpol: $|0\rangle$
- Südpol: $|1\rangle$

Jeder Punkt auf der Bloch Sphere kann durch zwei Winkel beschrieben werden:
- $\theta$ (Theta): Polarwinkel von der z-Achse ($0$ bis $\pi$)
- $\phi$ (Phi): Azimutwinkel in der xy-Ebene ($0$ bis $2\pi$)

Der zugehörige Zustand ist:
$$|\psi\rangle = \cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle$$

<div id="information" class="alert alert-success">
  <h2><i class="fas fa-info" style="font-size:36px"></i> &nbsp;  Informationen zur Bloch Sphere</h2>
    
  1. **$|0\rangle$:** $\theta=0$, $\phi=$ beliebig → Nordpol
  2. **$|1\rangle$**: $\theta=\pi$, $\phi=$ beliebig → Südpol</span>
  3. **$|+\rangle$**: $\theta=\pi/2$, $\phi=0$ → Punkt auf der x-Achse
  4. **$|-\rangle$**: $\theta=\pi/2$, $\phi=\pi$ → Gegenüber von $|+\rangle$ auf der x-Achse
</div>
