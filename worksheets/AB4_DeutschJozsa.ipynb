## Das Deutsch-Jozsa Problem

### Worum gehts?

Stell dir vor: Jemand stellt dir eine **geheime Funktion** zur Verfügung. Du weißt nur:

- Die Funktion hat als Eingabe eine Zahl (0, 1, 2, 3...)
- Die Funktion gibt immer **0 oder 1** zurück
- Die Funktion folgt einer **bestimmten Regel** und ist entweder *konstant* (es wird immer nur 0 oder 1 zurückgegeben) oder *balanciert* (es wird die gleiche Anzahl 0 und 1 zurückgegeben)

**Deine Aufgabe:** Finde heraus, welcher Regel die Funktion folgt!


**Klassisch** müsstest du viele Zahlen ausprobieren, um das Muster zu erkennen. Das kann unter Umständen sehr lange dauern.

Auf dem **Quantencomputer** geht das viel schneller! Wir benötigen nur **1 Versuch** (warum das so ist, finden wir später heraus)!


## Das Orakel-Konzept

Um dieses Problem zu lösen, nutzen wir das Konzept des **Orakels** als Werkzeug. Das Orakel ist die "Black Box", die die geheime Funktion versteckt.

Du kannst dir das **Orakel** also wie eine **Wissensbox** vorstellen. Eine Kiste, die du nicht öffnen kannst, aber der du **Fragen stellen** kannst und die immer eine (richtige) **Antwort** liefert.

![Orakel Übersicht](../figs/oracle_diagram.png)

**Beispiele aus dem Alltag:**
- 🔍 **Suchmaschine:** Du gibst ein Wort ein → bekommst Ergebnisse
- 🎱 **Magic 8-Ball:** Du schüttelst → bekommst eine Antwort
- 🤖 **KI-Assistent:** Du stellst eine Frage → bekommst eine Antwort

**In der Informatik** ist ein Orakel eine "Black Box" – wir kennen die Eingabe und Ausgabe, aber nicht, was genau in der Box passiert, um die Ausgabe zu erzeugen.

### Orakel im Quantencomputing

Im Quantencomputing nutzen wir Orakel als **Bausteine für Algorithmen**. Das Orakel:
- Nimmt einen **Bit-String** als Eingabe (z.B. 00, 01, 10, 11)
- Gibt **0 oder 1** als Ausgabe zurück
- Hat eine **bestimmte innere Regel**, die wir nicht kennen

### Das Deutsch-Jozsa Orakel

Beim **Deutsch-Jozsa Problem** gibt es nur **zwei mögliche Regeln** im Orakel:

| Typ | Was macht das Orakel? | Alltags-Beispiel | Ausgabe-Muster |
|-----|----------------------|-----------------|---------------|
| **Constant** | Gibt **immer das Gleiche** zurück | Ein Roboter, der immer "Ja" sagt | Immer 0... oder immer 1... |
| **Balanced** | Gibt **gleich häufig 0 und 1** zurück | Ein Münzwurf-Automat | 0, 1, 1, 0, 1, 0, 1, 0... |

### Die Herausforderung

**Wie oft musst du das Orakel fragen, um herauszufinden, welche Regel es verwendet?**

| Methode | Anzahl Fragen | |
|--------|--------------|--|
| **Klassisch** | Viele (exponentiell viele!) | 😰 |
| **Quantencomputing** | Nur **eine** einzige Frage! | 🎉 |

**Das ist der Quantenvorteil:** Was klassisch sehr aufwendig ist, schaffen Quantencomputer mit nur einem Schritt!

## 2) Der Deutsch-Jozsa Algorithmus

Der Deutsch-Jozsa Algorithmus löst das Problem mit nur **einer Abfrage**!

**Schritte:**
1. Alle Eingabe-Qubits in Superposition bringen (H-Gates)
2. Das Hilfs-Qubit in den Zustand $|-\rangle$ bringen (X-Gate + H-Gate)
2. Orakel anwenden
3. Nochmal H-Gates auf alle Eingabe-Qubits anwenden
4. Messen

<img src="../figs/DJ_small.png" alt="Überblick über den Deutsch-Jozsa Algorithmus mit einem Eingabe-Qubit" width="50%" />

**Ergebnis:**
- Wenn alle Mess-Qubit |0⟩ sind → Constant
- Wenn mindestens ein Mess-Qubit |1⟩ ist → Balanced

### Warum messen wir Qubit 1 und nicht Qubit 2?

**Das zweite Qubit (Qubit 2) ist der "Antwortzettel":**
- Es trägt die **Antwort** des Orakels (0 oder 1)
- Es ist im Zustand $|-\rangle$, um die Antwort zu "speichern"

**Aber wir lesen Qubit 1! Warum?**

Der Trick ist die **Interferenz** durch die H-Gates am Ende:

1. Nach dem Orakel steht die Antwort auf Qubit 2
2. Das H-Gate auf Qubit 1 am Ende sorgt für **Interferenz**
3. Die Information "springt" von Qubit 2 zu Qubit 1
4. Jetzt können wir das Ergebnis auf Qubit 1 ablesen!

**Warum |−⟩?**

Der Zustand $|-\rangle = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle)$ ist wichtig, weil:
- Er eine **Phase** enthält (das Minus)
- Diese Phase sorgt für **konstruktive oder destruktive Interferenz**
- Bei **Constant**: Alle Amplituden addieren sich → Ergebnis ist $|0\rangle$
- Bei **Balanced**: Amplituden löschen sich aus → mindestens ein Qubit ist $|1\rangle$

**Kurz gesagt:** Qubit 2 ist der "Speicher" für die Antwort, aber Qubit 1 ist der "Bildschirm", auf dem wir das Ergebnis sehen!

## 3) Warum funktioniert der Algorithmus?

Der Algorithmus nutzt zwei wichtige quantenmechanische Phänomene:

1. **Superposition:** Alle Eingaben werden gleichzeitig verarbeitet

2. **Interferenz:** Die H-Gates am Ende sorgen dafür, dass sich die Amplituden konstruktiv oder destruktiv überlagern

Bei Constant-Orakeln: Alle Amplituden addieren sich → Ergebnis ist |0..0⟩

Bei Balanced-Orakeln: Amplituden löschen sich aus → mindestens ein Qubit ist |1⟩
