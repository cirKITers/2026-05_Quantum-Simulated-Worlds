## Einleitung: Das Suchproblem

Stell dir vor, du hast eine unsortierte Liste mit $N$ Elementen und suchst nach einem bestimmten Element. Klassisch müsstest du im schlimmsten Fall $N$ Elemente durchsuchen - also $\mathcal{O}(N)$ Schritte.

Der Grover-Algorithmus schafft das in nur $\mathcal{O}(\sqrt{N})$ Schritten - eine **quadratische Beschleunigung**!

**Beispiel:**
- $N = 1.000.000$ Elemente
- Klassisch: Bis zu $1.000.000$ Versuche
- Grover: Nur $1.000$ Versuche!

## 1) Das Prinzip des Grover-Algorithmus

Der Grover-Algorithmus nutzt zwei wichtige Schritte:

1. **Orakel:** Das Orakel markiert die gesuchte Lösung (Amplitude wird negativ)

2. **Diffusion:** Die Diffusion verstärkt die Amplitude der markierten Lösung

Diese beiden Schritte werden mehrfach wiederholt, bis die gesuchte Lösung mit hoher Wahrscheinlichkeit gefunden wird.

```python
import random

# Klassische Suche
def klassische_suche(liste, ziel):
    """Sucht linear nach einem Element"""
    for i, element in enumerate(liste):
        if element == ziel:
            return i
    return -1

# Erstelle eine Liste
liste = random.sample(range(1, 9), 8)

# Suche nach dem Element 5
ziel = 5
index = klassische_suche(liste, ziel)

print(f"Liste: {liste}")
print(f"Suche nach: {ziel}")
print(f"Gefunden an Index: {index}")

# Wie viele Versuche brauchst du im Durchschnitt?
versuche = []
for _ in range(100):
    ziel = random.choice(liste)
    # Dein Code: Zähle die Versuche
    versuche.append(0)  # Platzhalter

print(f"\nDurchschnittliche Versuche (klassisch): {sum(versuche)/len(versuche):.2f}")
```

## 3) Warum und wie funktioniert der Grover Algorithmus?

Der Grover-Algorithmus nutzt **Interferenz**, um die Amplitude der gesuchten Lösung zu verstärken:

1. **Superposition:** Alle möglichen Lösungen werden gleichzeitig betrachtet

2. **Markieren:** Das Orakel macht die gesuchte Lösung "negativ"

3. **Verstärken:** Die Diffusion verstärkt die Amplitude der gesuchten Lösung

4. **Wiederholen:** Nach √N Iterationen ist die gesuchte Lösung mit hoher Wahrscheinlichkeit die Messung


### Die Geometrie von Grover

Eine Iteration des Grover-Algorithmus besteht aus **zwei Spiegelungen**. Zwei Spiegelungen hintereinander ergeben immer eine **Drehung**! Das ist wie bei einem Spiegelkabinett im Vergnügungspark.

#### Der 2D-Unterraum

Statt alle $2^n$ Zustände zu betrachten, reicht es, für die geometrische Herleitung nur **zwei Richtungen** zu betrachten:

- **$|w\rangle$ (gesucht):** Die Amplitude des gesuchten Zustands
- **$|r\rangle$ (restlich):** Die Summe aller anderen Amplituden

Der Zustandsvektor lebt in diesem 2D-Unterraum und wird bei jeder Iteration um den **gleichen Winkel** weitergedreht.

#### Die zwei Spiegelungen

| Schritt | Was passiert? | Geometrische Operation |
|---------|---------------|----------------------|
| **Orakel** | Macht die Amplitude von \|w⟩ negativ | Spiegelung an der \|r⟩-Achse |
| **Diffusion** | Verstärkt \|w⟩ und schwächt \|r⟩ ab | Spiegelung an der Mittelwert-Achse |

#### Warum funktioniert das?

1. Zu Beginn zeigen alle Amplituden in die gleiche Richtung (Superposition)
2. Das Orakel markiert das gesuchte Element und negiert damit die gesuchte Amplitude
3. Die Diffusion spiegelt alle Amplituden am Mittelwert und klappt damit die gesuchte Amplitude wieder nach oben
4. Nach jeder Iteration zeigt der Vektor ein bisschen mehr zum gesuchten Zustand
5. Nach $\sqrt{N}$ Schritten zeigt er fast genau auf $|w\rangle$!
