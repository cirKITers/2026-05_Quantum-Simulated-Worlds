## Was ist Verschränkung?

Verschränkung ist ein quantenmechanisches Phänomen, bei dem zwei oder mehr Qubits in einem gemeinsamen Zustand sind, sodass der Zustand eines Qubits nicht unabhängig von den anderen beschrieben werden kann.

Ein klassisches Analogon: Stell dir zwei Münzen vor, die durch einen magischen Mechanismus verbunden sind. Wenn du eine Münze wirfst und Kopf siehst, weißt du sofort, dass die andere Münze Zahl zeigt - egal wie weit sie voneinander entfernt sind.

In der Quantenmechanik ist dies noch seltsamer: Die Qubits befinden sich in einer Superposition, bis eine Messung durchgeführt wird.

## 1) Bell States

Bell States (auch EPR-Paare genannt) sind die einfachsten verschränkten Zustände zweier Qubits. Es gibt vier Bell States:

| Bell State | Formel | Beschreibung |
|------------|--------|---------------|
| Φ⁺ | (\|00⟩ + \|11⟩)/√2 | Gleicher Wert, gleiche Phase |
| Φ⁻ | (\|00⟩ - \|11⟩)/√2 | Gleicher Wert, ungleiche Phase |
| Ψ⁺ | (\|01⟩ + \|10⟩)/√2 | Ungleicher Wert, gleiche Phase |
| Ψ⁻ | (\|01⟩ - \|10⟩)/√2 | Ungleicher Wert, ungleiche Phase |

  <h3><i class="fa fa-laptop" style="font-size:28px"></i> <br><br> a) Bell State |Φ⁺⟩ erstellen  </h3>

  Erstelle den Bell State $|\phi^+\rangle = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$:

  1. Erstelle einen Schaltkreis mit 2 Qubits
  2. Wende das Hadamard-Gate an
  3. Wende das CNOT-Gate an
  4. Visualisiere den Zustand
  
  <hr>
      <i class="fas fa-wrench" style="font-size: 20px"></i> &nbsp; <strong>Variablen:</strong>  Verwende <code>qc.h()</code> und <code>qc.cx()</code>

```python
# Importiere Qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_state_city

# Erstelle Bell State |Φ⁺⟩
qc_bell = QuantumCircuit(2)

# Dein Code hier:
# 1. H-Gate auf Qubit 0
# 2. CNOT-Gate


# Erstelle den Zustandsvektor
state = Statevector.from_instruction(qc_bell)

# Zeige den Zustand
print("Zustandsvektor von |Φ⁺⟩:")
print(state)

# Visualisiere
print("\nVisualisierung:")
display(plot_state_city(state, title="Bell State |Φ⁺⟩"))
```

Ein wichtiger Unterschied zu dem, was wir uns in den Wochen zuvor angeschaut haben: Verschränkte Zustände können NICHT als einzelne Punkte auf separaten Bloch-Sphären dargestellt werden! Die Qubits haben keine individuellen reinen Zustände mehr.

Stattdessen müssen wir die gemeinsamen Zustandsvektoren visualisieren.

<div id="exercise" class="alert alert-info">
  <h3><i class="fa fa-laptop" style="font-size:28px"></i> <br><br> b) Messung von Bell States  </h3>
  Untersuche das Verhalten bei Messung:

  1. Erstelle den Bell State $|\phi^+\rangle$
  2. Führe 1000 Messungen durch
  3. Erkläre, warum die Ergebnisse korreliert sind
  4. Wiederhole für alle 4 Bell States
  
  <hr>
  <a href="../help/tipps_AB3_1b.ipynb"><i class="fas fa-life-ring" style="font-size:20px"></i> &nbsp;Tipp</a> <span style="margin-right: 20px;"></span>
  <a href="../help/zusatz_AB3_1b.ipynb"><i class="fas fa-puzzle-piece" style="font-size:20px"></i> &nbsp;Zusatzaufgabe</a>
</details>
</div>

```python
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# Erstelle und führe Messungen durch
def messung_bell(state_name, qc):
    """Führt Messungen auf einem Bell State durch"""
    qc.measure([0, 1], [1, 0])
    
    simulator = AerSimulator()
    job = simulator.run(qc, shots=1000)
    result = job.result()
    counts = result.get_counts()
    
    print(f"\n{state_name}:")
    print(f"Ergebnisse: {counts}")
    display(plot_histogram(counts))

# Bell State |Φ⁺⟩
qc_phi_plus = QuantumCircuit(2, 2)
# Dein Code: H und CNOT


messung_bell("|Φ⁺⟩ = (|00⟩ + |11⟩)/√2", qc_phi_plus)
```

```python
# Bell State |Φ⁻⟩ = (|00⟩ - |11⟩)/√2
qc_phi_minus = QuantumCircuit(2, 2)

# Dein Code hier:


messung_bell("|Φ⁻⟩ = (|00⟩ - |11⟩)/√2", qc_phi_minus)
```

```python
# Bell State |Ψ⁺⟩ = (|01⟩ + |10⟩)/√2
qc_psi_plus = QuantumCircuit(2, 2)

# Dein Code hier:


messung_bell("|Ψ⁺⟩ = (|01⟩ + |10⟩)/√2", qc_psi_plus)
```

```python
# Bell State |Ψ⁻⟩ = (|01⟩ - |10⟩)/√2
qc_psi_minus = QuantumCircuit(2, 2)

# Dein Code hier:


messung_bell("|Ψ⁻⟩ = (|01⟩ - |10⟩)/√2", qc_psi_minus)
```

<div id="information" class="alert alert-success">
  <h2><i class="fas fa-info" style="font-size:36px"></i> &nbsp;  Wichtige Eigenschaften von Bell States</h2>
    <ol>
        <li><span> <strong>Maximale Korrelation:</strong> Bei Messung sind die Ergebnisse immer korreliert</span></li>
        <li><span><strong>Nicht-Lokalität:</strong> Die Korrelation besteht, egal wie weit die Qubits voneinander entfernt sind</span></li>
        <li><span><strong>Keine separaten Zustände:</strong> Man kann den Zustand eines Qubits nicht unabhängig beschreiben</span></li>
    </ol>
</div>

# 2) 3-Qubit Verschränkung

Wir können nicht nur Verschränkungen über zwei Qubits machen, sondern beispielsweise auch über drei Qubits. Der sogenannte GHZ-Zustand bildet eine Verschränkung über drei Qubits ab und sieht folgendermaßen aus:

$$ |GHZ\rangle = \frac{1}{\sqrt{2}}(|000\rangle + |111\rangle) $$

<div id="exercise" class="alert alert-info">
  <h3><i class="fa fa-laptop" style="font-size:28px"></i> <br><br> a) GHZ-Zustand  </h3>
  Erstelle einen GHZ-Zustand (Greenberger-Horne-Zeilinger):
  
  1. Erstelle einen Schaltkreis mit 3 Qubits
  2. Wende H auf Qubit 0 an
  3. Überlege, wie du analog zum Bell-State $|\psi^+\rangle$, den GHZ-Zustand erzeugen kannst und wende die CNOT-Gates an
  4. Führe Messungen durch und erkläre das Ergebnis
  
  <hr>
    <a href="../help/tipps_AB3_2a.ipynb"><i class="fas fa-life-ring" style="font-size:20px"></i> &nbsp;Tipp</a> <span style="margin-right: 20px;"></span>
    <a href="../help/zusatz_AB3_2a.ipynb"><i class="fas fa-puzzle-piece" style="font-size:20px"></i> &nbsp;Zusatzaufgabe</a>
</div>

```python
# GHZ-Zustand erstellen
qc_ghz = QuantumCircuit(3, 3)

# Dein Code hier:


# Messung
qc_ghz.measure([0, 1, 2], [2, 1, 0])

simulator = AerSimulator()
job = simulator.run(qc_ghz, shots=1000)
result = job.result()
counts = result.get_counts()

print("GHZ-Zustand Messergebnisse:")
print(counts)
display(plot_histogram(counts))
```
